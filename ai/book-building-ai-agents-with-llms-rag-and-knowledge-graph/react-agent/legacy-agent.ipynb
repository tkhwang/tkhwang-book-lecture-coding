```python
from langchain_openai import ChatOpenAI
import dotenv

dotenv.load_dotenv()

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_community.tools import DuckDuckGoSearchRun, WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

llm = ChatOpenAI(model="gpt-4o-mini")  # 사용 모델로 교체

tools = [
    DuckDuckGoSearchRun(),
    WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper()),
]

```

```python
agent = create_agent(llm, tools, system_prompt="인터넷에서 검색하고, 위피키디아에서 윤석렬 대통령 검색해서 정리해서 알려줘.")

result = agent.invoke({"messages": [{"role": "user", "content": "2025년 한국 대통령은?"}]})

print(result["messages"][-1].content)
```
